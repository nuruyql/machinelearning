```python
import pandas as pd
import numpy as np

g = np.random.default_rng(0)
grades = pd.Series(
    g.integers(70,101,10),
    index = ['Sep','Oct','Nov','Dec','Jan','Feb','Mar','Apr','May','Jun','Jul']
)

print(grades)

year = grades.mean()
first = grades.iloc[:5].mean
second = grades.iloc[5:].mean


print(year)
print(first)
print(second)
```
