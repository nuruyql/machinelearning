```python
import pandas as pd
import matplotlib.pyplot as plt

donors = pd.read_csv("../data/donors.csv")

donors["size_bytes"].plot(kind="hist", bins=15, edgecolor="black")
plt.title("Distribution of agent sizes")
plt.xlabel("Size in bytes")
plt.ylabel("Number of agents")
plt.show()
```

```python
donors["author"].value_counts().head(10).plot(kind="bar")
plt.title("10 most common authors")
plt.xlabel("Author")
plt.ylabel("Number of agents")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

## Наблюдения по данным

1. Большинство записей относятся к агентам с `family_size = 1`; несколько более крупных значений выделяются на их фоне.
2. Среди 32 авторов 22 встречаются редко, поэтому данные распределены между авторами неравномерно.
3. В данных 31 категория `behaviour_family`, и 27 из них редкие. Это показывает, что набор включает много разных, но мало представленных типов поведения.
