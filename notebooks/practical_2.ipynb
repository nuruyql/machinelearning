## 9. Numerical Features Analysis

The numerical features are `family_size` and `size_bytes`. Their distributions and possible outliers are analyzed below.

```python
numeric_columns = ["family_size", "size_bytes"]

for column in numeric_columns:
    plt.figure(figsize=(8, 4))
    plt.hist(df[column].dropna(), bins=10, edgecolor="black")
    plt.title(f"Distribution of {column}")
    plt.xlabel(column)
    plt.ylabel("Frequency")
    plt.show()
```

### Outlier Detection

```python
for column in numeric_columns:
    plt.figure(figsize=(8, 3))
    plt.boxplot(df[column].dropna(), vert=False)
    plt.title(f"Boxplot of {column}")
    plt.xlabel(column)
    plt.show()
```

```python
for column in numeric_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ]

    print(f"\n{column}")
    print("Lower bound:", lower_bound)
    print("Upper bound:", upper_bound)
    print("Number of outliers:", len(outliers))
```

## 14. EDA Report and Data Quality Issues

1. **Missing values:** The dataset contains missing values in `payload_format`, `payload_b64`, and `verified`. On the next stage, these rows should be inspected to determine whether the missing values should be filled or kept as missing.

2. **Numerical distributions:** `family_size` has a very limited range of values, while `size_bytes` varies substantially. Their distributions should be considered before applying scaling or statistical models.

3. **Potential outliers:** The IQR and boxplot analysis identifies potential outliers in the numerical features. These observations should be inspected before deciding whether to keep, transform, or remove them.

4. **Rare categories:** Several categorical features contain categories that occur only once. Rare categories may need to be grouped or handled carefully before categorical encoding.

5. **High-cardinality features:** Columns such as `donor_id`, `source_url`, `main_py_sha256`, and `payload_b64` contain identifiers or highly unique information. They should generally not be used directly as ordinary categorical features in a machine-learning model.

6. **Correlation:** The numerical features `family_size` and `size_bytes` were analyzed for linear correlation. Because the dataset contains only two numerical features, the correlation analysis is limited and should not be overinterpreted.

7. **Target variable:** No explicit classification target is defined in the dataset. Therefore, class imbalance cannot be evaluated until a target variable is selected for a future modeling task.

```python
numeric_columns = ["family_size", "size_bytes"]

for column in numeric_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ]

    print(f"\n{column}")
    print("Q1:", Q1)
    print("Q3:", Q3)
    print("IQR:", IQR)
    print("Lower bound:", lower_bound)
    print("Upper bound:", upper_bound)
    print("Number of outliers:", len(outliers))
```

### Outlier Analysis

- `family_size` has 7 potential outliers according to the IQR method. However, Q1 and Q3 are both equal to 1, making the IQR equal to 0. Therefore, values greater than 1 are automatically classified as outliers. These values should not be removed automatically because they may represent valid larger families.
- `size_bytes` has no outliers according to the IQR method.
